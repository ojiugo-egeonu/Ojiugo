```python
import pandas as pd

df = pd.read_csv("state_data.csv")
```

```python
df.shape
```

```python
df.head()
df.tail()
median_lat = df["latitude"].median()
median_lon = df["longitude"].median()
df["lat_dist"] = abs(df["latitude"] - median_lat)
df["lon_dist"] = abs(df["longitude"] - median_lon)
closest_lat = df.loc[df["lat_dist"].idxmin()]
farthest_lat = df.loc[df["lat_dist"].idxmax()]

closest_lon = df.loc[df["lon_dist"].idxmin()]
farthest_lon = df.loc[df["lon_dist"].idxmax()]
```

```python
print(f"{closest_lat['state']} is closest: {closest_lat['lat_dist']:.2f}")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
df = sns.load_dataset("tips")  # example
df = df.dropna()
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# Top left (scatter)
axes[0,0].scatter(df["total_bill"], df["tip"])

# Top right (scatter)
axes[0,1].scatter(df["size"], df["tip"])

# Bottom left (histogram)
sns.histplot(df["total_bill"], ax=axes[1,0])

# Bottom right (boxplot)
sns.boxplot(x=df["day"], y=df["total_bill"], ax=axes[1,1])

plt.show()
```

```python
axes[0,0].set_title("Bills vs Tips")
axes[0,0].set_xlabel("Total Bill")
axes[0,0].set_ylabel("Tip")
```
