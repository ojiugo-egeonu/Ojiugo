```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

```python
df = pd.read_csv("noisy_curvefit_data.csv")
df.head()
```

```python
# shape
print(df.shape)

# rows and columns
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

# first 5
print(df.head())

# last 5
print(df.tail())

# column names
print(df.columns)
```

```python
plt.scatter(df["x"], df["y_noise_1"], color="gray", label="Noisy Data")

plt.xlabel("x")
plt.ylabel("y_noise_1")
plt.title("x vs y_noise_1")
plt.legend()

plt.show()
```

```python
def model(x, A, k):
    return A * np.exp(-k * x)

x = df["x"]
y_noise_1 = df["y_noise_1"]

from scipy.optimize import curve_fit

p0 = [1, 1]  # initial guess

popt, pcov = curve_fit(model, x, y_noise_1, p0=p0)

A_fit, k_fit = popt

print(f"Fit: y = {A_fit:.2f} * e^(-{k_fit:.2f} * x)")

residuals = y_noise_1 - model(x, A_fit, k_fit)
chi2 = np.sum(residuals**2)
dof = len(x) - len(popt)  # n - p
chi2_reduced = chi2 / dof

print(f"Reduced chi-square: {chi2_reduced:.2f}")

# smooth x for line
x_fit = np.linspace(min(x), max(x), 100)

# fitted model
y_fit = model(x_fit, A_fit, k_fit)

# true model (given in instructions)
y_true = model(x_fit, 5, 0.8)

plt.scatter(x, y_noise_1, color="gray", label="Noisy Data")
plt.plot(x_fit, y_fit, color="red", label="Fitted Model")
plt.plot(x_fit, y_true, linestyle="--", color="black", label="True Model")

plt.xlabel("x")
plt.ylabel("y")
plt.title(f"Noise Level 1 | Reduced χ² = {chi2_reduced:.2f}")
plt.legend()

plt.show()
```

```python
def fit_model(x, y):
    popt, _ = curve_fit(model, x, y, p0=[1,1])
    A_fit, k_fit = popt
    
    residuals = y - model(x, A_fit, k_fit)
    chi2 = np.sum(residuals**2)
    dof = len(x) - len(popt)
    chi2_reduced = chi2 / dof
    
    return A_fit, k_fit, chi2_reduced

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

noise_cols = ["y_noise_2", "y_noise_3", "y_noise_4", "y_noise_5"]

for i, col in enumerate(noise_cols):
    ax = axes[i//2, i%2]
    
    y = df[col]
    A_fit, k_fit, chi2_reduced = fit_model(x, y)
    
    x_fit = np.linspace(min(x), max(x), 100)
    
    ax.scatter(x, y, color="gray")
    ax.plot(x_fit, model(x_fit, A_fit, k_fit), color="red")
    ax.plot(x_fit, model(x_fit, 5, 0.8), linestyle="--", color="black")
    
    ax.set_title(f"{col} | χ² = {chi2_reduced:.2f}")
    ax.set_xlabel("x")
    ax.set_ylabel("y")

plt.tight_layout()
plt.show()
```

```python
results = []

for i in range(1, 6):
    col = f"y_noise_{i}"
    y = df[col]
    
    A_fit, k_fit, chi2_reduced = fit_model(x, y)
    
    results.append([i, A_fit, k_fit, chi2_reduced])

results_df = pd.DataFrame(results, columns=["Noise Level", "A_fit", "k_fit", "Reduced Chi^2"])

results_df.to_csv("noisy_parameters.csv", index=False)

results_df
```
