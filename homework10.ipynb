```python
import pandas as pd

data = {
    "Attribute": ["Mass", "Radius", "Gravity"],
    "Mercury": [0.33, 2439, 3.7],
    "Venus": [4.87, 6051, 8.9],
    "Earth": [5.97, 6371, 9.8],
    "Mars": [0.642, 3390, 3.7]
}

df = pd.DataFrame(data)
df.to_csv("solar_system.csv", index=False)
```

```python
df = pd.read_csv("solar_system.csv")
df.head()
```

```python
print("Before:", df.shape)

df_t = df.set_index("Attribute").T
df_t.index.name = "Planet"
df_t.reset_index(inplace=True)
df_t.columns.name = None

print("After:", df_t.shape)
```

```python
print("Before → rows:", df.shape[0], "columns:", df.shape[1])
print("After → rows:", df_t.shape[0], "columns:", df_t.shape[1])
```

```python
df1 = df.set_index("Attribute")
print("After set_index:", df1.shape)

df2 = df1.T
print("After transpose:", df2.shape)

df2.index.name = "Planet"
df2.reset_index(inplace=True)
df2.columns.name = None
print("Final shape:", df2.shape)
```

```python
def attach_units(df, old_col, new_col, unit):
    new_values = []
    
    for val in df[old_col]:
        if pd.notnull(val):
            new_values.append(val * unit)
        else:
            new_values.append(val)
    
    df[old_col] = new_values
    df.rename(columns={old_col: new_col}, inplace=True)
```
